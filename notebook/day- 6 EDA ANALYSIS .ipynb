```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(
    r"C:\Users\Dell\OneDrive\Desktop\ml project folder\creditcard.csv"
)
df_clean = df.copy()
print("Dataset shape:", df_clean.shape)
print("\nClass distribution:")
print(df_clean["Class"].value_counts())
df_clean["Class"].value_counts().plot(kind="bar")
plt.title("Genuine vs Fraud Transactions")
plt.xlabel("Class (0 = Genuine, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```

Univariate Analysis
- I analyzed the Class variable to understand the distribution of genuine and fraudulent transactions

```python
import matplotlib.pyplot as plt

print(df_clean["Class"].value_counts())

df_clean["Class"].value_counts().plot(kind="bar")

plt.title("Genuine vs Fraud Transactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean["Amount"], bins=50)

plt.title("Distribution of Transaction Amount")
plt.xlabel("Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df_clean["Time"], bins=50)

plt.title("Distribution of Transaction Time")
plt.xlabel("Time")
plt.ylabel("Frequency")

plt.show()
```

BIVARIATE ANALYSIS
- the relationship between two variables.

```python
plt.figure(figsize=(8,5))

df_clean.boxplot(column="Amount", by="Class")

plt.title("Transaction Amount by Class")
plt.suptitle("")
plt.xlabel("Class (0 = Genuine, 1 = Fraud)")
plt.ylabel("Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df_clean.boxplot(column="Time", by="Class")

plt.title("Transaction Time by Class")
plt.suptitle("")
plt.xlabel("Class")
plt.ylabel("Time")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df_clean.boxplot(column="V1", by="Class")

plt.title("V1 Distribution by Class")
plt.suptitle("")
plt.xlabel("Class")
plt.ylabel("V1")

plt.show()
```

MULTIVARIATE ANALYSIS
- Multi = more than two variables

```python
correlation = df_clean.corr(numeric_only=True)

plt.figure(figsize=(14,10))

import seaborn as sns

sns.heatmap(
    correlation,
    center=0
)

plt.title("Correlation Matrix")

plt.show()
```

```python
class_corr = correlation["Class"].sort_values()

print(class_corr)
```

```python
features = ["V1", "V2", "V3", "V4", "Amount", "Class"]

print(df_clean[features].groupby("Class").mean())
```
